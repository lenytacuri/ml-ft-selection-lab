# **What is Feature Selection?**

In the realm of machine learning, feature selection is the process of identifying and selecting a subset of relevant features for use in model construction. This process is crucial because the quality and quantity of features feeding into a model significantly impact its performance and interpretability. By choosing the right features, we can improve model accuracy, reduce complexity, and increase computational efficiency.

Imagine you're trying to predict the quality of wine based on various attributes (features) like acidity, sugar level, alcohol content, etc. Not all these attributes might be relevant to determining wine quality. Feature selection helps us identify which attributes (features) are essential and which can be ignored, thus simplifying our model without sacrificing accuracy.



# **Why is Feature Selection Important?**

Feature selection is not just about finding the needle in the haystack. It's about recognizing that not all data points are created equal when it comes to the predictive power they hold. Here are a few reasons why feature selection is vital:

1. **Improves Model Accuracy:** Eliminates noise by removing irrelevant or redundant features, thereby enhancing the model's predictive performance.
2. **Reduces Overfitting:** Fewer irrelevant features reduce the chance of making decisions based on noise, leading to a more generalized model.
3. **Increases Model Efficiency:** Less data means faster training. Models train quicker on fewer features.
4. **Enhances Model Interpretability:** A model with fewer features is easier to understand and explain, making the outcomes more actionable.

# **Curse of dimensionality**

The “curse of dimensionality” refers to a set of problems that arise when analyzing and organizing data in high-dimensional spaces (i.e., spaces with a large number of features or variables). As the number of dimensions increases, the volume of the space increases so fast that the available data become sparse, which presents challenges for statistical and machine learning methods.

⸻

🔍 **Key Problems Caused by the Curse of Dimensionality**

1.	Distance Metrics Become Less Informative: In high dimensions, the concept of “closeness” or “nearness” becomes meaningless. For example, the distance between all points tends to become similar, which negatively affects algorithms like k-NN, clustering, or any method relying on distance.

2.	Data Sparsity: With each added dimension, the volume of the feature space increases exponentially. This means more data is required to maintain the same density, but in practice, data is often limited. Thus, models trained in high dimensions risk overfitting.

3.	Increased Computational Cost: Algorithms operating in high-dimensional spaces require more memory and processing time, making them computationally expensive or infeasible.

4.	Model Overfitting: High-dimensional data allows models to fit noise rather than signal, especially when the number of dimensions exceeds the number of observations.

5.	Difficulty in Visualization and Interpretation:Human interpretation   impossible beyond three dimensions, making exploratory data analysis and intuition-based debugging hard.

⸻

🧠 **Examples**
* **Machine Learning:** In a dataset with hundreds or thousands of features (like image or gene data), models might perform worse unless dimensionality is reduced.

* **Search and Retrieval:** In high-dimensional vector spaces (e.g., embeddings in NLP), nearest neighbor searches become inefficient and unreliable.

* **Clustering:** Algorithms like k-means degrade in performance as distances lose meaning.

⸻

🛠️ **Common Solutions**
1.	Dimensionality Reduction:

* PCA (Principal Component Analysis)
* t-SNE, UMAP (for visualization)
* Autoencoders

2.	Feature Selection: Use only the most relevant features based on statistical methods or model performance.

3.	Regularization: Techniques like L1 (Lasso) or L2 (Ridge) can penalize unnecessary complexity and help reduce the effective number of features.


# **The Three Main Approaches to Feature Selection**

Feature selection techniques are broadly classified into three categories: **Filter Methods, Wrapper Methods, and Embedded Methods**. Each has its own merits and fits different scenarios.

image.png

## **Filter Methods: Unveiling Patterns through Statistical Scores**

Filter methods are a class of feature selection techniques that operate independently of any machine learning algorithm. Instead, they evaluate each feature based on certain statistical measures, and features are selected or discarded before introducing the data to the learning algorithm. This approach offers simplicity, speed, and efficiency in identifying relevant features. Let's explore some key aspects of filter methods:

1. **Statistical Scores:**
 * **Idea:** Filter methods rank features by their statistical scores, which measure the correlation or association between each feature and (in general) the target variable.
 * **Score (metrics):** Filter methods employ various statistical metrics, such as correlation coefficients, chi-square statistics, information gain, or mutual information, depending on the nature of the data (continuous or categorical) and the task at hand.
2. **Independence from Algorithms:**
 * **Advantage:** Filter methods are algorithm-agnostic. They assess features independently of the machine learning model that will eventually be applied. This makes them computationally less expensive and a suitable starting point for feature selection.
3. **Feature Ranking:**
 * **Process:** Features are ranked based on their individual scores. Those with higher scores are deemed more relevant, while lower-ranked features are considered less informative for predicting the target variable.
4. **Selection Criteria:**
 * **Thresholding:** A threshold is set, and features surpassing this threshold are retained, forming the selected subset.
5. **Advantages of Filter Methods:**
 * **Computational Efficiency:** Filter methods are faster compared to wrapper methods and embedded methods, making them suitable for datasets with a large number of features.
 * **Algorithm Independence:** Since they don't rely on a specific learning algorithm, filter methods are versatile and can be applied to various types of datasets and models.
6. **Disadvantages of Filter Methods:**
 * **Ignoring Feature Interactions:** Filter methods assess features in isolation and might overlook complex interactions between features. This limitation makes them less suitable for tasks where feature dependencies are crucial.
7. **Common Scores where target is required to compute the statistical relationship between each feature and the target:**
 * **Correlation Coefficient (for regression):** Measures linear relationships between continuous features and the target variable.
 * **Chi-Square Test (for classification):** Appropriate for categorical features, it assesses the independence of a feature from the target variable.
 * **Mutual Information (for both classification and regression):** Measures the dependence between two variables and is suitable for both continuous and categorical features.
 * **ANOVA F-test (for classification)**

8. **Common Scores where target is not required to compute the statistical relationship:**
* **Variance threshold:** Remove features with very low variance.
* **Correlation threshold:** Remove highly correlated features to avoid redundancy (feature-to-feature correlation).


## **Wrapper Methods: Iterative Search for Optimal Feature Sets**

Wrapper methods for feature selection treat the process as a search problem, involving the evaluation of different subsets of features using a predictive model. Unlike filter methods, wrapper methods actively involve a machine learning algorithm in the feature selection process. This iterative approach seeks to find the optimal set of features by training and evaluating the model multiple times. Here are the key aspects of wrapper methods:

1. **Search Strategies:**

 * **Idea:** Wrapper methods treat feature selection as a search problem, exploring different combinations of features to identify the subset that maximizes the performance of a given model.

 * **Examples:** Common search strategies include forward selection, backward elimination, and recursive feature elimination (RFE).

2. **Model Evaluation:**
 * **Criteria:** A predictive model is used to evaluate each subset of features, and a performance metric (e.g., accuracy, F1 score) is assigned to each combination.
 * **Scoring:** The quality of a feature subset is determined by the model's predictive performance on a validation set.
3. **Iterative Process:**
 * **Cycle:** Wrapper methods follow an iterative cycle of feature subset evaluation, model training, and refinement.
 * **Optimization:** The process continues until an optimal feature subset is found or until a specified stopping criterion is met.
4. **Feature Ranking:**
 * **Dynamic:** Features are dynamically selected or excluded in each iteration, allowing the algorithm to explore the entire feature space.
5. **Advantages of Wrapper Methods:**
 * **Model Sensitivity:** Wrapper methods consider the specific model's sensitivity to feature subsets, making them more suitable for complex relationships and interactions between features.

6. **Disadvantages of Wrapper Methods:**
 * **Computational Cost:** Wrapper methods are computationally more expensive than filter methods, as they require training the model multiple times.
 * **Overfitting Risk:** There's a risk of overfitting to the training data, especially if the dataset is small, due to multiple evaluations.
7. **Common Wrapper Methods:**
 * **Forward Selection:** Features are added one at a time, and at each step, the feature that provides the most improvement is included.
 * **Backward Elimination:** Starts with all features and removes the least significant one in each iteration until the specified stopping criterion is met.
 * **Recursive Feature Elimination (RFE)**: An iterative process where the least important features are pruned until the desired number of features is reached.

## **Embedded Methods: Inherent Feature Selection within Model Training**

Embedded methods seamlessly incorporate feature selection into the process of model training. Unlike filter and wrapper methods, embedded methods select features as a natural part of the algorithm's learning process. By doing so, they leverage the inherent capacity of certain algorithms to identify the most informative features. Here are the key aspects of embedded methods:

1. **Integration with Model Training:**
 * **Inherent Selection:** Features are selected or assigned weights during the model training process itself.
 * **Learning Process:** The algorithm optimizes both the model parameters and the importance of each feature simultaneously.

2. **Advantages of Embedded Methods:**
  * **Inherent Feature Interactions:** Embedded methods naturally consider interactions between features as part of the learning process, making them effective for capturing complex relationships.
  * **Model-Specific Selection:** Selection is tailored to the characteristics of the chosen model, ensuring a well-fitted subset of features.
3. **Disadvantages of Embedded Methods:**
 * **Computational Cost:** Some embedded methods, especially those involving iterative optimization, can be computationally expensive.
 * **Algorithm Dependency:** The effectiveness of embedded methods depends on the suitability of the chosen model for the specific dataset.
6. **Common Embedded Methods:**
 * **LASSO (Least Absolute Shrinkage and Selection Operator)**: Regularizes the linear regression model by adding a penalty term based on the absolute sum of feature coefficients.
 * **Tree-Based Models (Random Forest, Gradient Boosting):** Assigns importance scores to each feature during the ensemble learning process.

## **Baseline model using all available variables**
In our exploration of feature selection, we begin by loading and preparing the Breast Cancer dataset, a well-known benchmark in machine learning. This dataset encompasses various features related to breast cancer characteristics. To ensure robust model evaluation, we strategically split the dataset into training and testing sets. Subsequently, we standardize the features to bring them to a common scale, a crucial preprocessing step that aids in the fair comparison of their contributions to the model. With our data prepared, we proceed to implement and assess a baseline model, utilizing all available features. This initial model serves as our reference point, providing insights into the dataset's inherent characteristics and paving the way for the subsequent application of feature selection techniques to enhance model performance and interpretability.


### Revisión del modelo base y lectura de la tabla

- La separación usa `stratify=y` para conservar aproximadamente la proporción de malignos y benignos, y `random_state=42` para reproducirla.
- `Pipeline` une `StandardScaler` y `LogisticRegression`. Se ajusta con entrenamiento y se evalúa con test; el escalador no aprende del test. El código original también respetaba esta separación.
- Las etiquetas del dataset son **0 = maligno** y **1 = benigno**. La tabla calcula precision, recall y F1 para cada clase usando explícitamente `pos_label=0` y `pos_label=1`. El accuracy es global. Los guiones indican que la métrica no corresponde a esa fila.
- El resultado anterior, `0.9814814814814815`, era el **recall de la clase 1 (benigno)**: aproximadamente el 98.15 % de los benignos del test fueron clasificados como benignos. `recall_score` usa `pos_label=1` por defecto. Para medir cuántos malignos detecta el modelo, hay que mirar **Recall de clase 0**.
- Los resultados pueden cambiar respecto al código original porque la separación ahora es estratificada. Esta tabla es una evaluación con un único conjunto de test. Al ajustar hiperparámetros más adelante, el escalado y la selección de variables deben ajustarse dentro de cada partición de validación cruzada; el test se reserva para la evaluación final.

Referencia: [recall_score de scikit-learn](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.recall_score.html).


```python
# Librerías del laboratorio
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, f1_score, recall_score, roc_auc_score, precision_score

# Datos y etiquetas originales: 0 = maligno; 1 = benigno.
data = load_breast_cancer()
X = data.data
y = data.target
feature_names = data.feature_names
class_names = {0: "Maligno", 1: "Benigno"}

# Reservar el 30 % para test, conservando la proporción de ambas clases.
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

# El Pipeline aprende el escalado solo con train y lo aplica al predecir test.
baseline_model = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", LogisticRegression(max_iter=10000, random_state=42)),
])
baseline_model.fit(X_train, y_train)
y_pred_baseline = baseline_model.predict(X_test)


def metrics_table(y_true, y_pred):
    """Accuracy global y métricas por clase, en escala de 0 a 1."""
    rows = [{
        "Clase evaluada": "Global (0 y 1)",
        "Accuracy": accuracy_score(y_true, y_pred),
        "Precision": np.nan,
        "Recall": np.nan,
        "F1": np.nan,
        "Casos reales": len(y_true),
    }]
    for label, name in class_names.items():
        # pos_label indica qué clase se considera positiva en esta fila.
        rows.append({
            "Clase evaluada": f"{label} — {name} (pos_label={label})",
            "Accuracy": np.nan,
            "Precision": precision_score(y_true, y_pred, pos_label=label, zero_division=0),
            "Recall": recall_score(y_true, y_pred, pos_label=label, zero_division=0),
            "F1": f1_score(y_true, y_pred, pos_label=label, zero_division=0),
            "Casos reales": int(np.sum(np.asarray(y_true) == label)),
        })
    return pd.DataFrame(rows).set_index("Clase evaluada")


baseline_metrics = metrics_table(y_test, y_pred_baseline)
print(f"Modelo base: {X_train.shape[0]} casos de entrenamiento, "
      f"{X_test.shape[0]} de test y {X_train.shape[1]} variables.")
print("Métricas sobre test: escala 0–1 (0.95 equivale a 95 %).")
display(baseline_metrics.round(4).fillna("—"))
print("Recall 0: proporción de malignos reales detectados como malignos.")
print("Recall 1: proporción de benignos reales detectados como benignos.")

# Matrices escaladas que utilizan las siguientes secciones del laboratorio.
# Se reutiliza el escalador ya ajustado dentro del Pipeline.
scaler = baseline_model.named_steps["scaler"]
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# Matriz de confusión del modelo base sobre el conjunto de test.
import pandas as pd
from IPython.display import display, Markdown
from sklearn.metrics import confusion_matrix

# Convención: positivo = 1 (benigno); negativo = 0 (maligno).
# Filas = clase real; columnas = clase predicha.
# Con labels=[0, 1], el orden es [[TN, FP], [FN, TP]].
baseline_confusion_matrix = confusion_matrix(
    y_test, y_pred_baseline, labels=[0, 1]
)
tn, fp, fn, tp = baseline_confusion_matrix.ravel()

confusion_table = pd.DataFrame(
    [[f"TN = {tn}", f"FP = {fp}"],
     [f"FN = {fn}", f"TP = {tp}"]],
    index=["Real: 0 — Maligno (negativo)", "Real: 1 — Benigno (positivo)"],
    columns=["Predicho: 0 — Maligno (negativo)", "Predicho: 1 — Benigno (positivo)"],
)

display(Markdown("### Matriz de confusión — modelo base"))
display(Markdown("**Clase positiva: 1 = benigno. Clase negativa: 0 = maligno.**"))
display(confusion_table)
display(Markdown(
    f"- **TN (verdaderos negativos): {tn}** malignos clasificados como malignos.\n"
    f"- **FP (falsos positivos): {fp}** malignos clasificados como benignos.\n"
    f"- **FN (falsos negativos): {fn}** benignos clasificados como malignos.\n"
    f"- **TP (verdaderos positivos): {tp}** benignos clasificados como benignos.\n\n"
    "Aquí «positivo» significa pertenecer a la clase elegida (benigno), "
    "no tener cáncer. Si elegimos maligno como positivo, se invierten los papeles "
    "de TP/TN y FP/FN; las predicciones del modelo permanecen iguales."
))
```

## **Feature selection using a Filter Method**
The following code uses the `SelectKBest` function from the `sklearn.feature_selection` library in Python. This function is used to select the best features from a dataset based on a univariate statistical test.
In this case, the code is using the `mutual_info_classif` test to select the 10 best features from the `X_train_scaled` and `X_test_scaled` datasets. The selected features are then stored in the `X_train_k_best` and `X_test_k_best` variables, respectively.
The `mutual_info_classif` test measures the mutual information between each feature and the target variable. Mutual information is a measure of how much information one variable contains about another variable. In this case, the mutual information test is used to select the features that are most likely to be relevant to the target variable.
The `SelectKBest` function can be used to improve the performance of a machine learning model by reducing the number of features in the dataset. This can make the model more efficient and less likely to overfit the data.


### Datos utilizados y comparación con el modelo base

- `X_train_scaled` y `X_test_scaled` son las matrices preparadas con el escalador del modelo base, ajustado únicamente con entrenamiento.
- `y_train` y `y_test` contienen las clases **0 = maligno** y **1 = benigno**. No se estandarizan: son categorías, no magnitudes numéricas continuas. El filtro sí usa `y_train` en `fit_transform(X_train_scaled, y_train)` para medir la relación entre cada variable y la clase.
- El selector se ajusta solo con entrenamiento; `transform(X_test_scaled)` aplica al test las mismas 10 columnas. La semilla de información mutua queda fija para reproducir la selección.
- Se entrena una nueva regresión logística con la misma configuración del modelo base. La tabla reutiliza `metrics_table`, muestra ambas clases y compara exactamente los mismos casos de test.
- Ejecuta primero la celda del modelo base y luego esta celda. Así evitas mezclar resultados de ejecuciones anteriores con otra separación de datos. El valor `k=10` se mantiene como indica el ejercicio; esta comparación no se usa para elegir `k` sobre el test.


```python
### Método de filtro: información mutua ###
from functools import partial
from IPython.display import display, Markdown
from sklearn.base import clone
from sklearn.feature_selection import SelectKBest, mutual_info_classif

# X_train_scaled y X_test_scaled provienen del escalador del modelo base.
# y_train conserva las etiquetas: 0 = maligno y 1 = benigno. No se escala.
# La semilla hace reproducible la estimación de información mutua.
select_k_best = SelectKBest(
    score_func=partial(mutual_info_classif, random_state=42),
    k=10,
)

# Aprender qué variables seleccionar usando únicamente entrenamiento.
X_train_k_best = select_k_best.fit_transform(X_train_scaled, y_train)
# Aplicar esa misma selección al test, sin volver a ajustar el selector.
X_test_k_best = select_k_best.transform(X_test_scaled)
selected_features_filter = feature_names[select_k_best.get_support()]

display(Markdown(f"### Filtro: {len(selected_features_filter)} variables seleccionadas"))
display(pd.DataFrame({"Variable": selected_features_filter}))

# Misma configuración de regresión logística que el modelo base.
# clone crea un estimador nuevo, que se entrena con las 10 variables.
filter_model = clone(baseline_model.named_steps["classifier"])
filter_model.fit(X_train_k_best, y_train)
y_pred_filter = filter_model.predict(X_test_k_best)

# Comparar ambos modelos sobre los mismos casos y etiquetas de test.
filter_metrics = metrics_table(y_test, y_pred_filter)
filter_comparison = pd.concat(
    [baseline_metrics, filter_metrics],
    keys=[f"Base ({X_train.shape[1]} variables)",
          f"Filtro MI ({X_train_k_best.shape[1]} variables)"],
    names=["Modelo", "Clase evaluada"],
)

display(Markdown("### Comparación: modelo base vs. filtro de información mutua"))
display(Markdown(
    "Mismo conjunto de test. Valores entre **0 y 1** (0.95 = 95 %). "
    "**Accuracy** es global; **precision, recall y F1** se calculan por clase: "
    "**0 = maligno** y **1 = benigno**."
))
display(filter_comparison.round(4).fillna("—"))
```

This section uses matplotlib to create a bar chart that displays the importance of each feature visually.

```python
# Get feature importance scores
feature_scores = select_k_best.scores_

# Sort feature importance scores in descending order
indices = feature_scores.argsort()[::-1]

# Get the names of the sorted features
sorted_feature_names = feature_names[indices]

# Get the sorted feature importance scores
sorted_feature_scores = feature_scores[indices]

# Plot feature importance
plt.figure(figsize=(10, 6))
plt.bar(range(len(feature_names)), sorted_feature_scores)
plt.xticks(range(len(feature_names)), sorted_feature_names, rotation=90)
plt.xlabel("Features")
plt.ylabel("Importance")
plt.title("Feature Importance (Sorted)")
plt.tight_layout()
plt.show()
```

## **Feature selection using a Wrapper Method**

En esta sección usamos **RFE (Recursive Feature Elimination)** con regresión logística para conservar 10 de las 30 variables. RFE ajusta el modelo repetidamente y elimina en cada paso la variable con menor importancia según la magnitud de sus coeficientes. Al terminar, ajusta su estimador final con las variables seleccionadas.

### Preprocesamiento y evaluación

- Se reutilizan `X_train_scaled` y `X_test_scaled`, preparados con el escalador del modelo base y la misma separación estratificada. El escalador se ajustó únicamente con entrenamiento.
- `rfe.fit_transform(X_train_scaled, y_train)` aprende la selección con entrenamiento; `rfe.transform(X_test_scaled)` aplica esa misma selección al test. `y_train` conserva las etiquetas **0 = maligno** y **1 = benigno**, sin escalarlas.
- Este wrapper parte de las **30 variables originales escaladas**. Combinarlo con la salida del filtro corresponde al método híbrido de una actividad posterior.
- El estimador usa la misma configuración que el modelo base. `rfe.estimator_` ya está entrenado con las 10 variables finales y se utiliza para predecir.
- Se muestran las variables seleccionadas en su orden original (no es un ranking de importancia) y una tabla con una columna por modelo: **Base**, **Filtro MI** y **Wrapper RFE**. Las métricas se calculan sobre los mismos casos de test; precision, recall y F1 indican explícitamente la clase evaluada.

Ejecuta las celdas en orden: **modelo base → filtro → wrapper**. La cantidad de 10 variables está fijada por el ejercicio. Para elegirla mediante validación cruzada en otra etapa, el escalado y RFE deben ajustarse dentro de cada partición, reservando el test para la evaluación final.


```python
### Método wrapper: eliminación recursiva de variables (RFE) ###
from IPython.display import display, Markdown
from sklearn.base import clone
from sklearn.feature_selection import RFE

# Partir de las 30 variables escaladas del modelo base.
# RFE usa X_train_scaled y las etiquetas originales y_train (0/1).
# No utiliza las variables ya reducidas por el filtro de información mutua.
rfe = RFE(
    estimator=clone(baseline_model.named_steps["classifier"]),
    n_features_to_select=10,
    step=1,
)

# Seleccionar variables y ajustar el estimador exclusivamente con train.
X_train_rfe = rfe.fit_transform(X_train_scaled, y_train)
# Aplicar al test las columnas seleccionadas, sin volver a ajustar RFE.
X_test_rfe = rfe.transform(X_test_scaled)
selected_features_wrapper = feature_names[rfe.get_support()]

# RFE ya entrenó su estimador final con las 10 variables seleccionadas.
wrapper_model = rfe.estimator_
y_pred_wrapper = wrapper_model.predict(X_test_rfe)
wrapper_metrics = metrics_table(y_test, y_pred_wrapper)

display(Markdown(f"### Wrapper RFE: {len(selected_features_wrapper)} variables seleccionadas"))
display(pd.DataFrame({"Variable seleccionada": selected_features_wrapper}))


def metrics_column(table):
    """Convertir la tabla por clase en una columna para comparar modelos."""
    values = {"Accuracy global": table.loc["Global (0 y 1)", "Accuracy"]}
    for label, name in class_names.items():
        class_row = f"{label} — {name} (pos_label={label})"
        for metric in ["Precision", "Recall", "F1"]:
            values[f"{metric} — {label}: {name}"] = table.loc[class_row, metric]
    return pd.Series(values)


# Misma partición de test, mismas métricas y configuración de regresión logística.
model_comparison = pd.DataFrame({
    f"Base ({X_train.shape[1]} variables)": metrics_column(baseline_metrics),
    f"Filtro MI ({X_train_k_best.shape[1]} variables)": metrics_column(filter_metrics),
    f"Wrapper RFE ({X_train_rfe.shape[1]} variables)": metrics_column(wrapper_metrics),
})
model_comparison.index.name = "Métrica"

display(Markdown("### Comparación: modelo base, filtro y wrapper"))
display(Markdown(
    f"Evaluación sobre los mismos **{len(y_test)} casos de test**. "
    "Valores en **porcentaje**. Cada fila indica la clase evaluada: "
    "**0 = maligno** y **1 = benigno**."
))
# model_comparison conserva los valores numéricos 0–1 para reutilizarlos.
model_comparison_percent = model_comparison * 100
display(model_comparison_percent.map(lambda value: f"{value:.2f} %"))
```

## **Selección embedded: regresión logística con regularización L1**

Para clasificar **0 = maligno** y **1 = benigno**, utilizamos regresión logística con penalización L1. Su objetivo combina pérdida logística y penalización sobre la suma de valores absolutos de los coeficientes. L1 puede llevar coeficientes a cero, seleccionando variables durante el propio ajuste. Esta celda utiliza `GridSearchCV`, no `LassoCV` (que corresponde a regresión con pérdida cuadrática).

### Validación cruzada y preprocesamiento

- El `Pipeline` recibe `X_train` **sin escalar**. En cada una de las cinco particiones, `StandardScaler` aprende solo de las filas de entrenamiento de esa partición y transforma su validación. Así se evita filtrar información de validación al escalado.
- `StratifiedKFold` conserva aproximadamente las proporciones de malignos y benignos. Las etiquetas `y_train` no se escalan.
- Se prueban los valores de `C` de 0.20 a 0.29. Un menor `C` implica mayor regularización. El criterio de selección se declara explícitamente como **accuracy medio de validación**.
- `solver="saga"` optimiza los coeficientes; su tamaño de paso se calcula internamente. `max_iter=10000` es un límite por ajuste. En la versión instalada, `l1_ratio=1` especifica L1 pura y sustituye al parámetro obsoleto `penalty="l1"`.

### Modelo final y predicción

Con `refit=True`, la búsqueda elige el mejor `C` y ajusta un Pipeline nuevo con **todo el entrenamiento**. Las variables activas son las que tienen coeficientes distintos de cero en ese modelo final. No se promedian coeficientes ni se combinan listas de variables entre particiones.

`best_model.predict(X_test)` aplica el escalador final y utiliza los coeficientes L1 y el intercepto aprendidos. Las variables con coeficiente cero no contribuyen. Se pasan las 30 columnas originales porque el Pipeline espera esa estructura; no hace falta eliminar columnas físicamente para predecir. Se conservan todos los coeficientes no nulos, incluso los pequeños, para evaluar exactamente el modelo ajustado.

La antigua función `evaluate_model` creaba y entrenaba otra regresión logística y devolvía únicamente el recall de clase 1. Se ha eliminado: ahora `metrics_table(y_test, y_pred_l1)` calcula las métricas de las predicciones del modelo L1 final, sin reentrenarlo.

La tabla compara **base, filtro, wrapper y logística L1** sobre el mismo test, con accuracy global y precision, recall y F1 para ambas clases. L1 incluye selección de `C` mediante CV; los modelos anteriores mantienen su configuración fija. Ejecuta en orden **base → filtro → wrapper → L1**. El test se utiliza solo para evaluar.


```python
### Método embedded: regresión logística con regularización L1 ###
from IPython.display import display, Markdown
from sklearn.model_selection import GridSearchCV, StratifiedKFold

# L1 pura: l1_ratio=1 en la versión de scikit-learn de este proyecto.
# El Pipeline recibe datos SIN escalar para ajustar StandardScaler en cada fold.
l1_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", LogisticRegression(
        solver="saga", l1_ratio=1, max_iter=10000, random_state=42,
    )),
])

cv_l1 = StratifiedKFold(n_splits=5, shuffle=True, random_state=420)
param_grid = {"classifier__C": np.round(np.arange(0.20, 0.30, 0.01), 2)}
grid_search = GridSearchCV(
    estimator=l1_pipeline,
    param_grid=param_grid,
    scoring="accuracy",  # C se elige por accuracy medio de validación.
    cv=cv_l1,
    refit=True,  # Reajustar el Pipeline ganador con todo train, nunca con test.
    n_jobs=1,
    error_score="raise",
)
grid_search.fit(X_train, y_train)

# Modelo final: escalador + regresión L1 reajustados con todo entrenamiento.
best_params = grid_search.best_params_
best_model = grid_search.best_estimator_
l1_classifier = best_model.named_steps["classifier"]
coef = l1_classifier.coef_[0]

# SAGA puede producir ceros exactos: esas variables no aportan a la predicción.
# Se conserva todo coeficiente distinto de cero, sin modificar el modelo final.
selected_features = np.flatnonzero(coef != 0)
selected_features_l1 = feature_names[selected_features]
l1_selected_table = pd.DataFrame({
    "Variable seleccionada": selected_features_l1,
    "Coeficiente L1": coef[selected_features],
})

# Predecir con el Pipeline ya entrenado: aplica su escalador al test original.
# No se entrena otro clasificador ni se promedian coeficientes de los folds.
y_pred_l1 = best_model.predict(X_test)
l1_metrics = metrics_table(y_test, y_pred_l1)

# Matrices reducidas disponibles para actividades posteriores; no se necesitan
# para best_model.predict, que espera las 30 columnas originales en el mismo orden.
X_train_lasso = best_model.named_steps["scaler"].transform(X_train)[:, selected_features]
X_test_lasso = best_model.named_steps["scaler"].transform(X_test)[:, selected_features]

display(Markdown("### Modelo L1 final y variables seleccionadas"))
display(Markdown(
    f"**Mejor C:** {best_params['classifier__C']:.2f}. "
    f"**Accuracy medio de validación (5 folds):** {grid_search.best_score_:.2%}. "
    f"**Variables con coeficiente distinto de cero:** {len(selected_features)} de {X_train.shape[1]}. "
    f"**Iteraciones del ajuste final:** {l1_classifier.n_iter_[0]} "
    "(máximo permitido: 10000)."
))
display(l1_selected_table.round(6))

# Tabla acumulada: mismas métricas y los mismos casos de test para los 4 modelos.
model_comparison_l1 = pd.DataFrame({
    f"Base ({X_train.shape[1]} variables)": metrics_column(baseline_metrics),
    f"Filtro MI ({X_train_k_best.shape[1]} variables)": metrics_column(filter_metrics),
    f"Wrapper RFE ({X_train_rfe.shape[1]} variables)": metrics_column(wrapper_metrics),
    f"Logística L1 ({len(selected_features)} variables activas)": metrics_column(l1_metrics),
})
model_comparison_l1.index.name = "Métrica"
display(Markdown(f"### Comparación sobre los mismos {len(y_test)} casos de test"))
display(Markdown(
    "Valores en **porcentaje**; **0 = maligno**, **1 = benigno**. "
    "La columna L1 evalúa directamente el modelo regularizado elegido por CV. "
    "El test no se utiliza para elegir C ni las variables."
))
display((model_comparison_l1 * 100).map(lambda value: f"{value:.2f} %"))
```

# **Hybrid Methods**
Hybrid methods that combine elements of both filter and wrapper methods are often employed to harness the benefits of both approaches. These hybrid methods aim to leverage the efficiency of filter methods in quickly preselecting relevant features and the model-awareness of wrapper methods in fine-tuning the feature subsets. Here are a few hybrid methods:

* **Filter-Wrapper Hybrid**:
 * **Approach:** Initially, a filter method is used to preselect a subset of features based on some statistical measure (e.g., correlation, information gain). Then, a wrapper method is applied to fine-tune the selected subset using a predictive model.
 * **Benefits:** This approach aims to capitalize on the speed of filter methods for an initial feature selection and the model-awareness of wrapper methods to consider feature interactions.
* **Filter-Embedded Hybrid:**
 * **Approach:** Filter methods are used to preselect a subset of features based on some criteria. The selected features are then incorporated into an embedded method during model training. For example, features identified by a filter method may be used in LASSO regression.
 * **Benefits:** This hybrid approach combines the efficiency of filter methods with the model-awareness of embedded methods, allowing for simultaneous feature selection and model training.
* **Wrapper-Filter Hybrid:**
 * **Approach:** Similar to the filter-wrapper hybrid, a wrapper method is used to iteratively search for the best subset of features. Within each iteration, a filter method is applied to quickly evaluate feature subsets.
 * **Benefits:** This hybrid method aims to strike a balance between the exhaustive search of wrapper methods and the efficiency of filter methods, potentially reducing the computational cost.
* **Filter-Wrapper-Embedded Hybrid:**
 * **Approach:** This comprehensive hybrid method combines all three approaches. It starts with a filter method for initial feature selection, employs a wrapper method for further refinement, and integrates embedded methods during model training.
 * **Benefits:** By combining filter, wrapper, and embedded methods, this approach aims to capture the advantages of each, providing a well-rounded feature selection process.

Implementing hybrid methods requires careful consideration of the dataset characteristics, computational resources, and the specific goals of feature selection. The choice of hybrid method can be tailored based on the problem at hand and the trade-offs between computational efficiency and model interpretability.

**Objective:** The goal of this assignment is to understand and implement a Borda voting scheme to combine the results of multiple feature selection methods. **Understanding Borda Voting:**

Imagine you have a group of friends trying to decide on the best movie to watch. Each friend has their own ranked list of preferred movies. How do you combine these individual preferences into a single group ranking?
Borda voting is a simple and effective way to do this. It assigns points to each item (in our case, features) based on its position in each individual ranking. The item ranked highest by an individual receives the most points, the second highest receives the next most, and so on. The total Borda score for an item is the sum of the points it received from all individuals. The items with the highest total Borda scores are considered the group's top choices.
In the context of feature selection, each individual "friend" is a different feature selection method (Filter, Wrapper, Embedded). Each method provides a ranking of the features. Borda voting allows us to combine these rankings to get a consensus ranking of features.

**Your Task:**

4. **Embedded Method:** Random Forest Feature Importance (RandomForestClassifier).

Se recalculan sus coeficientes y se evalúa sobre el mismo test de 171 casos. `y_train` y `y_test` son etiquetas 0/1 y no se escalan. Al final aparece la comparación de accuracy global y precision, recall y F1 para **0 = maligno** y **1 = benigno**.
```python
### Borda voting: cuatro métodos recalculados, rankings desde 0 y top 5 ###
import numpy as np
import pandas as pd
from functools import partial
from IPython.display import display, Markdown
from sklearn.base import clone
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import SelectKBest, mutual_info_classif, RFE
from sklearn.linear_model import Lasso
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV, StratifiedKFold

n_features_borda = X_train.shape[1]
n_selected_features_borda = 5  # Fijado antes de evaluar; no se optimiza con test.
feature_indices_borda = np.arange(n_features_borda)

# 1. Recalcular información mutua: conserva 5, pero puntúa las 30 variables.
mi_model_borda = Pipeline([
    ("scaler", StandardScaler()),
    ("selector", SelectKBest(
        score_func=partial(mutual_info_classif, random_state=42),
        k=n_selected_features_borda,
    )),
])
mi_model_borda.fit(X_train, y_train)
mi_scores_borda = mi_model_borda.named_steps["selector"].scores_.copy()

# 2. Recalcular RFE hasta conservar 5 variables, con logística L2.
rfe_model_borda = Pipeline([
    ("scaler", StandardScaler()),
    ("selector", RFE(
        estimator=clone(baseline_model.named_steps["classifier"]),
        n_features_to_select=n_selected_features_borda,
        step=1,
    )),
])
rfe_model_borda.fit(X_train, y_train)
rfe_selector_borda = rfe_model_borda.named_steps["selector"]
# ranking_ es una agrupación nativa de sklearn (1 = las cinco sobrevivientes).
# Sus coeficientes finales están en el orden de las columnas que sobrevivieron.
rfe_native_ranking_borda = rfe_selector_borda.ranking_.copy()
rfe_abs_coef_borda = np.zeros(n_features_borda)
rfe_abs_coef_borda[rfe_selector_borda.get_support()] = np.abs(
    rfe_selector_borda.estimator_.coef_[0]
)
# Orden primario: ranking nativo. Desempate: mayor |coeficiente final|.
# Si dos magnitudes son idénticas, usar el índice original de columna.
# Las eliminadas conservan el orden de eliminación; no se inventan coeficientes.
rfe_order_borda = np.lexsort((
    feature_indices_borda, -rfe_abs_coef_borda, rfe_native_ranking_borda,
))

# 3. Recalcular Lasso de regresión con validación cruzada.
# GridSearchCV + Pipeline(Lasso) ajusta el escalado dentro de cada fold.
# Lasso ordena las 30 variables; no se fuerza a tener exactamente 5 coeficientes.
lasso_cv_borda = StratifiedKFold(n_splits=5, shuffle=True, random_state=420)
lasso_search_borda = GridSearchCV(
    estimator=Pipeline([
        ("scaler", StandardScaler()),
        ("lasso", Lasso(max_iter=100000, tol=1e-5)),
    ]),
    param_grid={"lasso__alpha": np.logspace(-4, 0, 40)},
    scoring="neg_mean_squared_error",
    cv=lasso_cv_borda,
    refit=True,
    n_jobs=1,
    error_score="raise",
)
lasso_search_borda.fit(X_train, y_train)
lasso_model_borda = lasso_search_borda.best_estimator_
lasso_coef_borda = lasso_model_borda.named_steps["lasso"].coef_.copy()
lasso_importance_borda = np.abs(lasso_coef_borda)

# 4. Recalcular las importancias de Random Forest sobre entrenamiento.
forest_model_borda = Pipeline([
    ("scaler", StandardScaler()),
    ("forest", RandomForestClassifier(
        n_estimators=300, max_features="sqrt", random_state=42, n_jobs=1,
    )),
])
forest_model_borda.fit(X_train, y_train)
forest_importance_borda = forest_model_borda.named_steps["forest"].feature_importances_.copy()

# 5. Convertir cada lista ORDENADA de índices en el RANGO de cada variable.
# Ejemplo: orden [2, 0, 1] → rangos [1, 2, 0]; la variable 2 ocupa el puesto 0.
def zero_based_ranks(order):
    order = np.asarray(order, dtype=int)
    if not np.array_equal(np.sort(order), np.arange(len(order))):
        raise ValueError("El orden debe contener cada índice exactamente una vez.")
    ranks = np.empty(len(order), dtype=int)
    ranks[order] = np.arange(len(order))
    return ranks


# En MI, Lasso y RF, mayor importancia es mejor. Los valores exactamente
# iguales (incluidos ceros Lasso) se ordenan por índice original, sin usar test.
# Ese desempate residual es determinista; no implica una diferencia de evidencia.
mi_order_borda = np.argsort(-mi_scores_borda, kind="stable")
lasso_order_borda = np.argsort(-lasso_importance_borda, kind="stable")
forest_order_borda = np.argsort(-forest_importance_borda, kind="stable")
borda_rankings = pd.DataFrame({
    "MI": zero_based_ranks(mi_order_borda),
    "RFE": zero_based_ranks(rfe_order_borda),
    "Lasso CV": zero_based_ranks(lasso_order_borda),
    "Random Forest": zero_based_ranks(forest_order_borda),
}, index=pd.Index(feature_names, name="Variable"))

# 6. Puntos Borda: n_features - rank. Con 30 variables: 30, 29, ..., 1.
# Los cuatro métodos tienen el mismo peso; ya no se promedian rangos empatados.
borda_points = n_features_borda - borda_rankings
borda_scores = borda_points.sum(axis=1).to_numpy()
# Un empate en el total se resuelve por el orden original de las columnas.
borda_order = np.argsort(-borda_scores, kind="stable")
borda_selected_indices = borda_order[:n_selected_features_borda]
selected_features_borda = feature_names[borda_selected_indices]
borda_points_table = borda_points.copy()
borda_points_table.columns = [f"Puntos {method}" for method in borda_points.columns]
borda_points_table["Total Borda"] = borda_scores

# 7. Mismos pacientes de train/test, ahora con las 5 columnas elegidas por Borda.
X_train_borda = X_train[:, borda_selected_indices].copy()
X_test_borda = X_test[:, borda_selected_indices].copy()
y_train_borda = y_train.copy()
y_test_borda = y_test.copy()
borda_model = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", clone(baseline_model.named_steps["classifier"])),
])
borda_model.fit(X_train_borda, y_train_borda)
y_pred_borda = borda_model.predict(X_test_borda)
borda_metrics = metrics_table(y_test_borda, y_pred_borda)

borda_selected_table = pd.DataFrame({
    "Variable seleccionada": selected_features_borda,
    "Total Borda": borda_scores[borda_selected_indices],
    "Coeficiente del modelo final": borda_model.named_steps["classifier"].coef_[0],
}, index=pd.Index(np.arange(n_selected_features_borda), name="Posición Borda (desde 0)"))

# 8. Mostrar las tres tablas solicitadas y las métricas comparativas.
display(Markdown("### Borda: cuatro métodos recalculados para seleccionar 5 variables"))
display(Markdown(
    "MI: **k=5**; RFE: **5 variables, logística L2, step=1**, "
    "desempate por magnitud de coeficientes finales; "
    "Random Forest: **300 árboles, semilla 42**. "
    f"Lasso: **alpha={lasso_search_borda.best_params_['lasso__alpha']:.6g}**, "
    f"{np.count_nonzero(lasso_coef_borda)} coeficientes no nulos; "
    "5 folds con escalado interno. Los cuatro métodos ordenan las 30 variables."
))
display(Markdown("### Tabla 1. Ranking de cada variable por método: 0 = mejor, 29 = última"))
with pd.option_context("display.max_rows", None, "display.max_columns", None):
    display(borda_rankings.iloc[borda_order])
display(Markdown("### Tabla 2. Puntos por método: 30 − ranking; total = suma de los cuatro"))
with pd.option_context("display.max_rows", None, "display.max_columns", None):
    display(borda_points_table.iloc[borda_order])
display(Markdown("### Tabla 3. Las 5 variables elegidas, sus puntos y sus coeficientes finales"))
display(borda_selected_table.round(6))

# Construir desde los cinco modelos anteriores evita conservar una columna
# Borda antigua o duplicarla al volver a ejecutar esta celda.
model_comparison_borda = model_comparison_hybrid.copy()
model_comparison_borda[
    f"Borda ({n_selected_features_borda} variables)"
] = metrics_column(borda_metrics)
display(Markdown(f"### Comparación final — los mismos {len(y_test_borda)} casos de test"))
display(Markdown(
    "Valores en **porcentaje**. Accuracy global; precision, recall y F1 "
    "para **0 = maligno** y **1 = benigno**. "
    "El test no participa en el escalado, rankings, desempates ni selección."
))
display((model_comparison_borda * 100).map(lambda value: f"{value:.2f} %"))
```

4. Can you think of any potential disadvantages or limitations of using Borda voting for feature selection?

- Igual peso para métodos de distinta calidad: en la implementación anterior, todos votan igual, aunque alguno resulte menos adecuado para estos datos.
- Redundancia: varias variables muy correlacionadas pueden recibir muchos votos y desplazar a otras que aporten información complementaria.
- Sensibilidad a decisiones del procedimiento: el resultado depende de los métodos incluidos, los desempates y el número final de variables.
- Mayor costo de entrenamiento: requiere ajustar varios selectores, algunos con múltiples entrenamientos o validación cruzada.
- No garantiza mejor predicción: una buena posición por consenso no implica que ese conjunto maximice el recall, la precisión o el F1 del modelo final.
